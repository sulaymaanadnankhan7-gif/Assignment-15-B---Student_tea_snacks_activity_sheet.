```python
tea_counts = students["tea_type"].value_counts()
tea_counts
```

```python
most_preferred_tea = tea_counts.idxmax()
most_preferred_tea
```

```python
average_cups_by_gender = students.groupby("gender")["cups_per_day"].mean()
average_cups_by_gender
```

```python
snack_counts = students["snack_choice"].value_counts()
snack_counts
```

```python
snack_counts.plot(kind="bar", color="green", width=0.5, xlabel="Snack Choice", ylabel="Number of Students", title="Snack Choices", rot=45)
```

```python
tea_by_age_group = pd.crosstab(students["age_group"], students["tea_type"])
tea_by_age_group
```

```python
tea_by_age_group.plot(kind="bar", xlabel="Age Group", ylabel="Number of Students", title="Tea Preference by Age Group", rot=0)
```
